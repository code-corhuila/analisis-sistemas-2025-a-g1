# Análisis Ampliado de la Encuesta a Clientes

Este notebook permite visualizar los requerimientos o funcionalidades más importantes del proyecto a partir de los datos de la encuesta a clientes.
Se presentan diversos gráficos para una mejor interpretación de la información.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

# Cargar el archivo Excel
file_path = 'Instrumentos_Recoleccion_Informacion.xlsx'
clientes_df = pd.read_excel(file_path, sheet_name='Encuesta Clientes')

# Mostrar las primeras filas de los datos
clientes_df.head()
```

## 1. Distribución de la Frecuencia de Compra

```python
# Contar las respuestas de la frecuencia de compra
frecuencia_contador = clientes_df['Frecuencia de compra'].value_counts()

# Graficar
plt.figure(figsize=(8, 5))
sns.barplot(x=frecuencia_contador.index, y=frecuencia_contador.values, palette='viridis')
plt.xlabel('Frecuencia de compra')
plt.ylabel('Cantidad de respuestas')
plt.title('Distribución de la Frecuencia de Compra')
plt.xticks(rotation=45)
plt.show()
```

## 2. Preferencia de Plataforma para Compras en Línea

```python
# Contar las respuestas sobre la plataforma preferida
plataforma_contador = clientes_df['Sistema preferido'].value_counts()

# Graficar
plt.figure(figsize=(8, 5))
sns.barplot(x=plataforma_contador.index, y=plataforma_contador.values, palette='coolwarm')
plt.xlabel('Sistema preferido')
plt.ylabel('Cantidad de respuestas')
plt.title('Preferencia de Plataforma para Compras en Línea')
plt.xticks(rotation=45)
plt.show()
```

## 3. Servicios Preferidos en la Aplicación

```python
# Procesar datos de los servicios preferidos
servicios_lista = clientes_df['Servicios preferidos'].dropna().str.split(', ').sum()
servicios_contador = Counter(servicios_lista)

# Graficar
plt.figure(figsize=(10, 5))
sns.barplot(x=list(servicios_contador.values()), y=list(servicios_contador.keys()), palette='magma')
plt.xlabel('Frecuencia')
plt.ylabel('Servicios preferidos')
plt.title('Servicios más solicitados por los clientes')
plt.show()
```

## 4. Importancia de las Notificaciones

```python
# Contar las respuestas sobre la importancia de las notificaciones
notificaciones_contador = clientes_df['Importancia de notificaciones'].value_counts()

# Graficar
plt.figure(figsize=(8, 5))
sns.barplot(x=notificaciones_contador.index, y=notificaciones_contador.values, palette='Set2')
plt.xlabel('Importancia de las notificaciones')
plt.ylabel('Cantidad de respuestas')
plt.title('Opinión sobre la Importancia de las Notificaciones')
plt.xticks(rotation=45)
plt.show()
```
